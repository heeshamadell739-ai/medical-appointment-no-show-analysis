Data Manipulation / Wrangling

```python
# Feature engeenering 
# Cratea WaitingDays coulmn 
med_App['WaitingDays'] = (med_App['AppointmentDay'] -med_App['ScheduledDay']).dt.total_seconds() / (24 * 60 * 60)
med_App['WaitingDays'] = med_App['WaitingDays'].round().astype(int)
med_App['WaitingDays'].describe()
```

```python
#Investigate negative WaitingDays
med_App[med_App['WaitingDays'] < 0]
```

```python

# Day column
med_App['AppointmentWeekday'] = med_App['AppointmentDay'].dt.day_name()
# MedicalConditionCount Column
med_App['MedicalConditionCount'] = (med_App['Hipertension'].astype(int)+ med_App['Diabetes'].astype(int)+ med_App['Alcoholism'].astype(int)+ (med_App['Handcap'] > 0).astype(int))
```

```python
# Mapping IsNoShow column values to 0 , and 1 , For mean calculation 
med_App['IsNoShow'] = (med_App['No-show'].map({'No': 0, 'Yes': 1}).astype(int))

#Dividing age into groups 
bins = [-1, 9, 19, 29,39,49,59,69,79, float('inf')]
labels = ['0 -9','10-19','20-29','30 -39','40-49','50-59','60 -69','70-79','80+',]
med_App['Age_Group'] = pd.cut(med_App['Age'], bins=bins,labels=labels)

#Dividing waiting days into groups
med_App = med_App[med_App['WaitingDays'] >= 0].copy()
bins = [-1, 0, 7, 30, 90, float('inf')]
labels = ['Same day','1-7 days','8-30 days','31-90 days','90+ days']
med_App['WaitingCategory'] = pd.cut(med_App['WaitingDays'],bins=bins,labels=labels)
```

# Question 1:
# Does the waiting time between scheduling and the appointment
# affect no-show rates?

```python
med_App['WaitingCategory'].value_counts().sort_index()
```

```python
# Calculate the no-show rate and count for each waiting-time category
waiting_analysis = (med_App.groupby('WaitingCategory', observed=True)['IsNoShow'].agg(["mean","count"]).reset_index())
# Convert the no-show proportion to a percentage
waiting_analysis['NoShowRate'] = (waiting_analysis['mean'] * 100).round(2)
# Remove the intermediate mean column
waiting_analysis.drop(columns='mean', inplace=True)


waiting_analysis
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=waiting_analysis,x="WaitingCategory",y="NoShowRate")


plt.title('No-show Rate by Waiting Time')
plt.xlabel('Waiting Time')
plt.ylabel('No-show Rate (%)')
```

```python
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=waiting_analysis,
    x='WaitingCategory',
    y='count'
)

plt.title('Number of Appointments by Waiting Time')
plt.xlabel('Waiting Time')
plt.ylabel('Number of Appointments')
```

Insight

 No-show rates increase as the waiting period becomes longer, with patients who have longer waiting periods being more likely to miss their appointments.

This suggests that the higher overall no-show rate among SMS recipients is likely influenced by differences in waiting time between the two groups.

Overall Conclusion
The analysis indicates that no-show behavior is influenced by multiple factors rather than a single characteristic. Among the investigated variables, waiting time shows one of the clearest relationships with attendance, with longer waiting periods generally associated with higher no-show rates. The analysis also suggests that SMS reminders may be more beneficial for appointments with longer waiting periods.
Neighbourhoods show noticeable differences in no-show rates, with some areas performing substantially above or below the overall no-show rate of 22.08%. Age groups also show some variation, particularly with younger patients having relatively higher no-show rates, although the differences are not large enough to consider age alone a strong explanation.
Appointment weekday shows relatively small differences in no-show rates, suggesting that the day of the appointment is not a major factor. Similarly, the number of recorded medical conditions does not show a clear positive relationship with no-shows.
Overall, the findings suggest that appointment lead time and patient/location characteristics may be more useful for identifying higher-risk appointments than simply looking at the appointment day or medical-condition count.

Final Recommendations
1. Prioritize patients with long waiting periods
Patients whose appointments are scheduled several weeks in advance should receive additional reminders or confirmation requests, as longer waiting periods are associated with higher no-show rates.

2. Use SMS reminders strategically
SMS reminders appear particularly useful for appointments with waiting periods of 8–90 days. The hospital could prioritize reminder campaigns for these appointments rather than treating every appointment equally.

3. Investigate high-risk neighbourhoods
Neighbourhoods with substantially higher no-show rates should be investigated further to identify possible underlying factors such as transportation difficulties, accessibility, socioeconomic conditions, or differences in healthcare access.

4. Target younger patient groups
Younger age groups, particularly 10–29, show relatively higher no-show rates. Additional reminders or confirmation procedures could be considered for these groups.

5. Don't overreact to small groups
Decisions should be based on sufficiently large patient populations. Results from groups with very few appointments, such as the 4-condition medical group or Saturday appointments, should not be used as the basis for operational decisions.

6. Track both No-show Rate and No-show Volume
Management should monitor both the percentage of patients who miss appointments and the absolute number of missed appointments. A group with a moderate no-show rate but a very large patient population can generate more missed appointments than a small group with a very high no-show rate.
